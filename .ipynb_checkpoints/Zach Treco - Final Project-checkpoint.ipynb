## Introduction

The experiment that this final project is based on is one that measures star flux over time for the star KIC 8462852 in the Cygnus constellation, located approximately $1480$ light years from Earth. The project serves as an analysis of this data to find whether the measured flux of this star is significantly increasing or decreasing. Flux, in the context of photons, is the quantitative measure of the energy that radiates on some unit of area per unit of time. The standard unit of energy flux is Joules per square meter per second ($J/m^2s$).

This collection of data from the Mikulski Archive for Space Telescopes (MAST) was created with 53 full-frame images (FFI) from different space telescopes. Using the images, the normalized flux of KIC 8462852 was calculated for each image and reported to the dataset. Each calculation is tagged with the number of days after a Barycentric Julian Date (BJD) for reference.

Montet et al. (2017) thoroughly explains the process that the space telescope _Kepler_ underwent to capture its own FFIs.

##### Hypothesis Creation:

This project will statistically determine whether the flux data is significantly different throughout the dataset. Visual analysis shows a clear decrease in flux over time; however, one must determine whether this decrease is a statistically significant decrease when considering other factors such as the uncertainty in each flux calculation. Therefore, I propose the null hypothesis that the flux of star KIC 8462852 does **not** significantly change over time. The alternative hypothesis is that the flux of star KIC 8462852 does significantly change (either increase or decrease) over time.

##### Note:

The process of calculating the normalized flux of each data point is explained in Monet (2016), but it's a bit complicated. To my understanding, the star's "aperature photometry" is compared to reference stars that are used for calibration. I assume that these stars have a known flux that is quite consistent and thus can be used as experimental controls. Each image's pixels are analyzed for their standard deviations and the measured flux of each data point is normalized relative to the mean value. This is how the error column of the data was created.

## Graph of Dataset

##### Fitted to an exponentially decreasing logged linear regression

To determine whether the data may have an exponential fit, we'll use a weighted first degree polynomial fit. Plotting the log of the y-values should return a linear regression of the data if it fits to a natural exponential function. Note that numpy.log is the natural log, ln, so the inverse is an exponential function of the form $Ae^x$.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

# Import data
time = np.loadtxt("PHYS267_DataSet_Astrophysics_StarFlux_Full.csv", usecols=0, skiprows=1, delimiter=',')
flux = np.loadtxt("PHYS267_DataSet_Astrophysics_StarFlux_Full.csv", usecols=1, skiprows=1, delimiter=',')
sigma = np.loadtxt("PHYS267_DataSet_Astrophysics_StarFlux_Full.csv", usecols=2, skiprows=1, delimiter=',')
orientation = np.loadtxt("PHYS267_DataSet_Astrophysics_StarFlux_Full.csv", usecols=3, skiprows=1, delimiter=',')

pars = np.polyfit(time, np.log(flux), 1, w=sigma)               # Set parameters, degree of fit, and weights to account for error
print("The degree 1 polynomial parameters are",pars)
print()
model = np.poly1d(pars)                                 # Creates a model as printed based on the best fit parameters from pars
model_vals = model(time)                                # Finds the model values for each x-value (distance) to calculate the R^2 score
print("The model for this degree is")
print(model)
print()
    
# Plot the velocity vs. distance graph with respective degree of best-fit
print("The following plot is for a degree 1 semi-logged function:")
plt.errorbar(time, np.log(flux), yerr=sigma, fmt='.', ecolor='green', elinewidth=0.6, \
    capsize=2)
plt.plot(time, model_vals, '-')
plt.title("Measured Flux of Star KIC 8462852 starting at Barycentric Julian Date-2454833 ($ln$($y$-values))")
plt.xlabel("Time (Days after Barycentric Julian Date-2454833)")
plt.ylabel("$ln$ of Normalized Flux")
plt.show()
    
print("The R^2 score for this model is",r2_score(np.log(flux), model_vals))
```

## Model Fitting

##### Creating a linear regression of the data

```python
import numpy as np
import matplotlib.pyplot as plt

degree = np.arange(1,21,1)  # Create some range of degrees that is definitely more than needed for the fit

for i in degree:
    pars = np.polyfit(time, flux, i, w=sigma)               # Set parameters, degree of fit, and weights to account for error
    print("The degree",i,"polynomial parameters are",pars)
    print()
    model = np.poly1d(pars)                                 # Creates a model as printed based on the best fit parameters from pars
    model_vals = model(time)                                # Finds the model values for each x-value (distance) to calculate the R^2 score
    print("The model for this degree is")
    print(model)
    print()
    
    # Plot the velocity vs. distance graph with respective degree of best-fit
    print("The following plot is for degree",i)
    plt.errorbar(time, flux, yerr=sigma, fmt='.', ecolor='green', elinewidth=0.6, \
        capsize=2)
    plt.plot(time, model_vals, '-')
    plt.title("Measured Flux of Star KIC 8462852 starting at Barycentric Julian Date-2454833 (No error bars)")
    plt.xlabel("Time (Days after Barycentric Julian Date-2454833)")
    plt.ylabel("Normalized Flux")
    plt.show()
    
    print("The R^2 score for this model is",r2_score(flux, model_vals))
    print()
    # If statement to determine when a suitable degree of best-fit is reached
    if r2_score(flux, model_vals) >= 0.95:
        print("Degree",i,"meets the R^2 requirement of at least 0.95 \
and is a suitable fit for the data.")
        break
    else:
        print("*******************************************************************************")
        print()
        continue
```

Looking at each degree of the model, we see a general downtrend in the data. As the degrees increase, the tail end of the data looks like it should trend back upwards if we were to collect more data. It then suddenly predicts a further decrease in flux if you increase the degree even further.

Though the model passes the chosen $R^2$ score at the eigth degree, I will argue the seventh degree model as more suitable because the model curve is continuous across the whole dataset. The eigth degree curve has cusps, which are not good for a physical model since there is no derivative at these points.

#### First Data Analysis Test:
##### Moments of the Data Distribution

```python
import numpy as np
from scipy.stats import skew, kurtosis

# I'm going to use np.average to calculate the weighted average of the data along with the mean. This takes error into account

# Find the moments of distribution
mean = np.mean(flux)   # Calculates mean
print("Mean:", mean)
avg = np.average(flux, weights=sigma)
print("Weighted Average:", avg)
var = np.var(flux)     # Calculates variance
print("Variance:", var)
skew = skew(flux)      # scipy.stats.skew() gives the skewness of a distribution or dataset
print("Skew:", skew)
kurt = kurtosis(flux)  # scipy.stats.kurtosis() gives the *EXCESS KURTOSIS* of a dataset
print("Kurtosis:", kurt + 3) # Always add 3 if using kurtosis() so you get the actual kurtosis
```

#### Second Data Analysis Test:
##### Normal Test

It's important to determine whether this data is normally distributed or not since this will determine the types of Hypothesis tests we can run on the data. This will also give better context for model fitting.

```python
from scipy.stats import stats

# Perform normal test
stat, pNorm = stats.normaltest(flux)
if pNorm > 0.05:
    print("The data is confirmed distributed normally with a p-value of",pNorm)
else:
    print("The data is confirmed NOT distributed normally with a p-value of",pNorm)
```

The p-Value above give a decimal probability that the data is normally distributed. Since it is only a $1.4\%$ chance that it is, we can safely conclude that the data is **not** distributed normally with almost $99\%$ confidence.

#### Third Data Analysis Test:
##### One-Sample Wilcoxon Signed Rank Sum Test 

The normal test leads us to the second data analysis test, a hypothesis test to determine whether the data is significantly different across the dataset.

We now know that:
* The data is non-parametric (not normally distributed)
* There is only one dataset being analyzed
* We need to compare each data point to a reference value in the dataset

Using the One-Sample Wilcoxon Signed Rank Sum Test, one can determine how significant the flux values differ from a reference value. For this, we will use the normalized mean of $1$. This will help us determine whether the range of flux measurements is statistically different.

When performing this test, we will use a significance level of $\alpha = 0.05$. This will give us a suitable $95\%$ confidence level if our p-Value turns out higher than $0.95$. The null and alternate hypotheses are the same as those stated in the introduction.

```python
from scipy.stats import wilcoxon
average = 1
difference = [data - average for data in flux]
wrank, wrank_pVal = wilcoxon(difference)
print("Rank:", wrank, "| p-Value:", wrank_pVal)

if wrank_pVal < 0.05:
    print("Since p-value < 0.05, we can reject the null hypothesis; \
concluding there may be a difference in the data. This will be especially likely \
if we draw the same conclusion from other tests.")
else:
    print("Since p-value > 0.05, we accept the null hypothesis; \
concluding there is probably no difference in the data. This will be especially likely \
if we draw the same conclusion from other tests.")
```

#### Fourth Data Analysis Test:

##### Critical Value and P-Value Test

We will compare data at the far ends of the dataset to help determine if there is a significant difference in the data.

The easiest way to do this is to take the mean value of the data and compare it with the two test statistics at either end of the dataset.

I decided to try an exponential fit for this test since we already know that the data is not normally distributed. However, the data somewhat follows an exponential decay visually, so let's try it:

```python
import scipy.stats

# Set up variables for the first statistic test
mu = np.mean(flux)
xbar_lower = flux[0]
xbar_upper = flux[len(flux)-1]
s = np.sqrt(np.var(flux))
n = len(flux)

# We will use a significane level of alpha = 0.05 for both tests

alpha = 0.05
zcrit = scipy.stats.expon.ppf(1-alpha/2) # /2 for a two-sided test
print("The critical z-score is:",zcrit)
print()
zscore_lower = abs((xbar_lower-mu)/(s/np.sqrt(n)))
print("The z-score of the earliest data point, which is also one of the highest flux values, is:",zscore_lower)
print("The corresponding p-value is:",scipy.stats.expon.sf(zscore_lower)*2)
print("Therefore, we have a confidence level of:",1-scipy.stats.expon.sf(zscore_lower)*2)
print()
zscore_upper = abs((xbar_upper-mu)/(s/np.sqrt(n)))
print("The z-score of the latest data point, which is also one of the lowest flux values, is:",zscore_upper)
print("The corresponding p-value is:",scipy.stats.expon.sf(zscore_upper)*2)
print("Therefore, we have a confidence level of:",1-scipy.stats.expon.sf(zscore_upper)*2)
```

## Addressing Possible Errors in Data

##### Determining whether the orientation of the telescope had a significant effect on the data and finding further evidence to support whether we accept or reject the null hypothesis.

Part of the dataset lists the orientation that the camera was in when collecting that particular image. Montet, et al. (2017), explains that the _Kepler_ space telescope had to rotate its orientation every 93 days to reposition its heat shield as it orbited the Sun. Therefore, there may be a discrepancy when comparing data between different orientations in the dataset.

To test this, I shall separate the data by orientation and plot the data separately to see whether each orientation is statistically similar when independent of each other.

```python
import numpy as np
import matplotlib.pyplot as plt

o0 = np.array([])
o1 = np.array([])
o2 = np.array([])
o3 = np.array([])
o0time = np.array([])
o1time = np.array([])
o2time = np.array([])
o3time = np.array([])

for i in np.arange(0,len(flux),1):
    if orientation[i] == 0:
        o0 = np.append(o0, flux[i])
        o0time = np.append(o0time, time[i])
    elif orientation[i] == 1:
        o1 = np.append(o1, flux[i])
        o1time = np.append(o1time, time[i])
    elif orientation[i] == 2:
        o2 = np.append(o2, flux[i])
        o2time = np.append(o2time, time[i])
    elif orientation[i] == 3:
        o3 = np.append(o3, flux[i])
        o3time = np.append(o3time, time[i])
    else:
        continue

print("The data with orientation = 0:")
print(o0)
print()
print("The data with orientation = 1:")
print(o1)
print()
print("The data with orientation = 2:")
print(o2)
print()
print("The data with orientation = 3:")
print(o3)
```

To gain a bit of insight into whether the orientation had a significant effect on the data collection, I'll run a few Mann-Whitney tests since the two datasets are independent and non-parametric.

```python
import scipy.stats

mw_u, mw_pVal = scipy.stats.mannwhitneyu(o0, o2, alternative='two-sided')
print("o0, o2")
print("U Statistic:", mw_u, "| p-Value:", mw_pVal)
print()
mw_u, mw_pVal = scipy.stats.mannwhitneyu(o1, o3, alternative='two-sided')
print("o1, o3")
print("U Statistic:", mw_u, "| p-Value:", mw_pVal)
print()
mw_u, mw_pVal = scipy.stats.mannwhitneyu(o0, o3, alternative='two-sided')
print("o0, o3")
print("U Statistic:", mw_u, "| p-Value:", mw_pVal)
print()
mw_u, mw_pVal = scipy.stats.mannwhitneyu(o1, o2, alternative='two-sided')
print("o1, o2")
print("U Statistic:", mw_u, "| p-Value:", mw_pVal)
print()
mw_u, mw_pVal = scipy.stats.mannwhitneyu(o0, o1, alternative='two-sided')
print("o0, o1")
print("U Statistic:", mw_u, "| p-Value:", mw_pVal)
print()
mw_u, mw_pVal = scipy.stats.mannwhitneyu(o2, o3, alternative='two-sided')
print("o2, o3")
print("U Statistic:", mw_u, "| p-Value:", mw_pVal)
```

Notice that the value of these tests only fall below 0.05 once out of these combinations. Most of the combinations of data subsets return p-Values that do not have a reliable confidence level for rejecting the null hypothesis. Therefore, this data generally supports acceptance of the null hypothesis that the flux is statistically similar (the same) throughout the dataset. 

We also conclude that the changing orientation of the _Kepler_ space telescope liekly had no statistically significant effect on the measurement of flux.

## Discussion

Based on the various statistical analyses conducted, there are various suggested conclusions that we can draw. The model fitting exercise gave us insight into where the data is trending towards and whether the star may be continuously decreasing in flux for a significant time after the end of the dataset. Lower degree fits, especially degrees 2 and 3, put the data on an exponential-like downtrend. It’s also worth noting that taking the log of the flux values to create a semi-log plot made no difference in the best-fit line relative to the data. It also did not affect the $R^2$ score at all, suggesting that the data is not actually exponential in nature. The chosen $R^2$ score threshold was 0.95. This wasn’t achieved until an 8th degree polynomial. However, such a large degree makes predictions about the behavior of the data difficult without running actual calculations and creating further plots. When fitting models, we tend to prefer lower-degree fits since these simplify complex observations in nature. Despite this, the high degree models to seem to show that the data will trend back upwards, which may support the null hypothesis that the data is not significantly changing.

It's always good to calculate the moments of a distribution since it allows us to make assumptions about the data’s behavior and trends. No matter whether the data is normally distributed or not, there are conclusions that one can draw based on the variance, skew, and kurtosis of the data. Though the normal test ruled out a normal distribution, we assume a normal distribution when calculating the moments of the distribution.

As one may observe, the variance of the data seems to be extremely tiny. This is due to analyzing a dataset that only has a range of about 0.035 from its highest and lowest value. Thus, the standard deviation is about 33% of the range of the entire dataset. This implies that the data is spread far apart in the dataset, which is possible considering that much of the data is concentrated at the upper and lower bounds of the dataset. This may indicate that the data is statistically different from each other within the dataset, but this will be better discussed when discussing more reliable tests.
The skew of the data is negative, which means that the data is concentrated more to the right half of the graph and tails off towards the left. The magnitude of the skew is between 0.5 and 1, which means that the skew is moderate and can be easily seen visually on the data histogram.
Considering the kurtosis of the graph at 2.39, we have a kurtosis less than that of a normal distribution. The “peak” of the distribution is thus Platykurtic with a very domed peak. The tails also tend to be “taller” than that of a normal distribution.

Overall, we can confidently support that out data is not normally distributed. However, we conducted a normal test to be sure whether we should use parametric or non-parametric hypothesis test.

The normal test was necessary to distinguish whether the sample data set could be modeled to a larger, normally distributed population. The normal test concluded with approximately 99% confidence that our data is not normal. This guided which tests we would be able to perform on this data later in the project.

Since the primary objective of this project is to determine whether the data is significantly changing, we tested it with the Wilcoxon Signed Rank Sum Test using one sample. We have a 69% confidence that the null hypothesis is correct and thus should conclude that the data is likely similar across the dataset. The “dip” in the data may be insignificant and only a minor fluctuation on a small scale.

To find more support towards one of our hypotheses, the critical value/p-value tests compared data at both ends of the dataset to the average relative flux of about 0.98. Since the variance and standard deviation of the sample were quite small, this led to small differences in data values to have high z-scores. The extremes of the data fell at anywhere from 6σ to 12σ. This test leads to the conclusion that the data are significantly different across the dataset. This conflicts with the Wilcoxon test, but the null hypothesis shouldn't suddenly be rejected based on this test. It wouldn't be good to reject the null and risk a false positive when other tests suggest that the null may be true. Note that this test only compares single points and does not put all of the data in context. Thus, more testing is necessary before making a conclusion.

This led to a third comparison test in which many combinations of the data, separated by orientation, were directly compared to each other using the Mann-Whitney test. The tests concluded that there were generally no differences between the subsets of data. One test did return a 95.2% confidence in rejecting the null hypothesis; however, we cannot ignore that other combinations of data were nowhere close to this confidence level. Therefore, these Mann-Whitney tests lead to further support the acceptance of the null hypothesis that the data is likely statistically similar.

## Conclusion

Though our statistical tests may not 100% conclude whether we should accept or reject the null hypothesis that the data in this dataset is similar, the prevailing conclusion is to accept the null hypothesis. 

It is also generally more dangerous to reject a null hypothesis when it is actually true (Type I Error) than to accept it when it is actually false (Type II Error). Therefore, there is not high enough confidence in our calculations to safely reject the null hypothesis when looking at the entire dataset. When conducting the critical values test, points at the extremes of the data may show that those singular data points are much different, but it does not take the entire dataset into consideration. This is a non-parametric dataset with a small variance, so the tests may conclude that the data is less similar than reality.

The other two tests, the Signed Rank Sum and Mann-Whitney tests both give more compelling evidence that suggests that the null hypothesis is true. The first of which returns a 69% probability that the null hypothesis is true after comparing each flux value to the relative flux of 1. The Mann-Whitney test further supports this by giving low confidence levels in rejecting the null hypothesis, except for one comparison between two data subsets.

Even the fitted models trend towards an increase in flux towards the end of the data. This may further support that the data is unchanging and could be proven with more data collection after the ~1600 day mark.

Overall, the data, though it looks to be almost exponentially trending lower, is not significantly changing. There is high confidence that there is no considerable difference in the data after fitting models and conducting several statistical tests.
